```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df.columns = [
    col.strip().replace(" ","_")
    for col in df.columns
]
```

```python
df.columns
```

```python
df["Purchase_Date"].dtype
```

```python
df["Order_ID"] = [
    f"OrderID{i}" for i in range(1, len(df) + 1)
]
```

```python
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])
```

```python
df["Purchase_Date"].dtype
```

```python
df["Purchase_Time"] = pd.to_datetime(df["Purchase_Time"],format="%H:%M:%S").dt.time
```

```python
df["Age_Group"] = pd.cut(
    df["Customer_Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=[
        "Under 18",
        "18-30",
        "31-45",
        "46-60",
        "60+"
    ]
)
```

```python
df["Year"] = df["Purchase_Date"].dt.year

df["Month"] = df["Purchase_Date"].dt.month

df["Month_Name"] = df["Purchase_Date"].dt.month_name()

df["Quarter"] = df["Purchase_Date"].dt.quarter

df["Day_Name"] = df["Purchase_Date"].dt.day_name()
```

```python
df["Hour"] = pd.to_datetime(
    df["Purchase_Time"],
    format="%H:%M:%S"
).dt.hour
```

```python
df.head()
```

```python
dim_product = df[
    [
        "Product_Name",
        "Category",
        "Price"
    ]
].drop_duplicates()
```

```python
dim_product["Product_ID"] = range(
    1,
    len(dim_product) + 1
)
```

```python
dim_customer = df[
    [
        "Customer_ID",
        "Customer_Age",
        "Customer_Gender",
        "Age_Group"
    ]
].drop_duplicates()
```

```python
dim_date = pd.DataFrame({
    "Date": df["Purchase_Date"].drop_duplicates()
})

dim_date["Year"] = dim_date["Date"].dt.year

dim_date["Month"] = dim_date["Date"].dt.month

dim_date["Month_Name"] = (
    dim_date["Date"].dt.month_name()
)

dim_date["Quarter"] = (
    "Q" + dim_date["Date"].dt.quarter.astype(str)
)

dim_date["Day"] = dim_date["Date"].dt.day

dim_date["Day Name"] = (
    dim_date["Date"].dt.day_name()
)
```

```python
product_lookup = dim_product[
    ["Product_ID", "Product_Name", "Category", "Price"]
].copy()
df = df.merge(
    product_lookup,
    on=["Product_Name", "Category", "Price"],
    how="left"
)
df["Product_ID"].isna().sum()
```

```python
df.to_csv("Ecommerce_sales_fact.csv", index=False)
```
